# COVID-19 Data Analysis (Canada)

A Python time-series and exploratory data analysis portfolio project using a **simulated** daily COVID-19 dataset.

> **Data disclosure:** The included data is synthetic/simulated for learning and visualization practice. It is not official Canadian COVID-19 surveillance data and must not be presented as real reported cases, deaths, hospitalizations, or vaccination figures.

## Objectives
- Explore daily case, death, hospitalization, and vaccination trends.
- Calculate rolling averages to smooth daily variation.
- Compare trends over time.
- Examine correlations among simulated variables.
- Explain limitations of simulated pandemic data.


## 1. Import libraries

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

pd.set_option("display.max_columns", None)
```

## 2. Load and inspect the dataset

```python
df = pd.read_csv("covid19_canada_simulated.csv", parse_dates=["date"])
df = df.sort_values("date").reset_index(drop=True)

print("Dataset shape:", df.shape)
display(df.head())
display(df.describe())
print("\\nMissing values:")
print(df.isnull().sum())
```

## 3. Daily cases over time

```python
plt.figure(figsize=(12, 5))
plt.plot(df["date"], df["daily_cases"], linewidth=1)
plt.title("Simulated Daily COVID-19 Cases in Canada")
plt.xlabel("Date")
plt.ylabel("Daily cases (simulated)")
plt.tight_layout()
plt.show()
```

## 4. Seven-day rolling average

```python
df["cases_7day_avg"] = df["daily_cases"].rolling(window=7, min_periods=1).mean()

plt.figure(figsize=(12, 5))
plt.plot(df["date"], df["daily_cases"], alpha=0.35, label="Daily cases")
plt.plot(df["date"], df["cases_7day_avg"], linewidth=2, label="7-day average")
plt.title("Daily Cases and 7-Day Rolling Average (Simulated)")
plt.xlabel("Date")
plt.ylabel("Cases")
plt.legend()
plt.tight_layout()
plt.show()
```

## 5. Deaths and hospitalizations

```python
fig, axes = plt.subplots(2, 1, figsize=(12, 8), sharex=True)
axes[0].plot(df["date"], df["daily_deaths"])
axes[0].set_title("Simulated Daily Deaths")
axes[0].set_ylabel("Deaths")
axes[1].plot(df["date"], df["hospitalizations"])
axes[1].set_title("Simulated Hospitalizations")
axes[1].set_ylabel("Hospitalizations")
axes[1].set_xlabel("Date")
plt.tight_layout()
plt.show()
```

## 6. Vaccination trend

```python
plt.figure(figsize=(12, 5))
plt.plot(df["date"], df["vaccination_rate_percent"])
plt.title("Simulated Vaccination Rate Over Time")
plt.xlabel("Date")
plt.ylabel("Vaccination rate (%)")
plt.ylim(0, 100)
plt.tight_layout()
plt.show()
```

## 7. Correlation analysis

```python
numeric_cols = [
    "daily_cases", "daily_deaths", "hospitalizations",
    "vaccination_rate_percent"
]
correlation = df[numeric_cols].corr()
display(correlation.round(3))

plt.figure(figsize=(7, 5))
plt.imshow(correlation, aspect="auto")
plt.colorbar(label="Correlation")
plt.xticks(range(len(numeric_cols)), numeric_cols, rotation=35, ha="right")
plt.yticks(range(len(numeric_cols)), numeric_cols)
plt.title("Correlation Matrix (Simulated Data)")
plt.tight_layout()
plt.show()
```

## 8. Monthly summary

```python
df["month"] = df["date"].dt.to_period("M").astype(str)
monthly = df.groupby("month", as_index=False).agg({
    "daily_cases": "sum",
    "daily_deaths": "sum",
    "hospitalizations": "mean",
    "vaccination_rate_percent": "mean"
})
display(monthly.head(12))
```

## 9. Conclusion and limitations

Use the plots and monthly summary generated above to describe the simulated waves and trends. Do not interpret these values as real Canadian pandemic statistics.

This dataset is simulated for programming practice. It cannot support conclusions about actual COVID-19 transmission, mortality, hospital burden, vaccine effectiveness, or public policy.
